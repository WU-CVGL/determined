# JupyterLab in Determined

This JupyterLab is the `jupyter` service of a Determined shell. It runs in the shell's container, with the shell's environment and resources.

## Persistence

Save your work to a shared filesystem: the files in the container go when the shell ends.

## Idle Timeout

A shell whose configuration sets `idle_timeout` ends once it has been idle that long.
JupyterLab is active as `services.jupyter.idle_type` defines it: by default, while a kernel is executing.
